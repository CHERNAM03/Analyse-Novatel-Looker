Nous te présenterons un nouveau format de fichier de données. Tu connais déjà les fichiers CSV ou d'autres formats de fichiers comme .xlsx, mais nous souhaitons parfois représenter les données dans une structure arborescente. Pour ce faire, nous pouvons utiliser des fichiers JSON.

#### Arborescence (tree structure)

![](https://www.tutorialspoint.com/data_structures_algorithms/images/binary_tree.jpg)

En informatique, un arbre est une structure de données composée de nœuds. Un nœud est une structure qui peut contenir une valeur ou une condition, ou représenter une structure de données distincte (qui pourrait être un arbre en soi).

Chaque nœud d'un arbre a zéro ou plusieurs nœuds enfants, qui se trouvent en dessous de lui dans l'arbre (par convention, les arbres sont dessinés en poussant vers le bas). Un nœud qui a un enfant est appelé nœud parent de l'enfant (ou supérieur). Un nœud a au plus un parent, mais éventuellement plusieurs nœuds ancêtres, comme le parent du parent.

Les nœuds enfants ayant le même parent sont des nœuds frères et sœurs.

On trouve beaucoup de structures arborescentes en informatique, par exemple le répertoire et le fichier sur un ordinateur suivent une structure arborescente, même chose avec une page HTML. Il peut également être puissant pour représenter des données.

![image](https://storage.googleapis.com/quest_editor_uploads/ppqLn65Fdz9sWTSk5cRogcdQGaCWpCvp.jpg)


#### JSON

JSON (JavaScript Object Notation) est un format permettant de transcrire des données en texte. Ce format est de plus en plus utilisé par les développeurs. Les données sont classées en objets et sous la forme d'une structure arborescente :

- une racine a une clé (le nom de l'objet)
- ses branches ont des clés et des valeurs (les attributs de l'objet)

Voici à quoi ressemble un objet JSON :

```json
{
    "student": {
        "firstname": "Eric",
        "lastname": "Cartman",
        "age": 10,
        "psychopath": true,
        "emergencyCall": {
            "firstname": "Liane",
            "lastname": "Cartman",
            "number": 5556754
        },
        "hobbies": [
            "World of Warcraft",
            "Tacos",
            "Drink tears"
        ],
        "friends": null
    }
}

```

##### Objet JSON

Un objet JSON a la forme suivante. Ca ressemble à un objet que tu connais déjà en python : un dictionnaire.

```json
{
    key1: value1,
    key2: value2
}

```

##### Array JSON

Un array JSON a la forme suivante. Ca ressemble à un objet que tu connais déjà en python : une liste.

```json
key1: [
    value1,
    value2
]
```

Un tableau est représenté par des crochets : à une clé est associé un ensemble de valeurs.

#### JSON et Python

Un fichier JSON, comme un fichier CSV, est un fichier texte. L'extension du nom de fichier nous sert simplement à savoir comment l'interpréter.
Tu peux utiliser la bibliothèque `json` pour interpréter (on dit "parser") ce texte par python.

```python
import json

text = """{
    "key1": "value1",
    "key2": "value2"
     }"""

data = json.loads(text)
```

Ce texte est alors interprété par python comme des dictionnaires et des listes imbriqués. Tu peux donc maintenant t'en servir comme tu as l'habitude :

```python
data['key2']
```

```shell-black
>>> 'value2'
```

Il faut donc jongler entre les listes et les dictionnaires imbriqués. Sur le premier exemple en haut de cette page, si on veut afficher chacun des hobbies, il faut donc faire :

```python
for hobby in data['student']['hobbies']:
  print(hobby)
```


```shell-black
>>> World of Warcraft
>>> Tacos
>>> Drink tears
```

### Quel est le format d'une API REST ?

Une API est une interface d'échange entre plusieurs langages. Une RESTful API, ou API REST, est une API avec un formalisme spécifique, notamment :
- La requête s'effectue avec le protocole HTTP (le même que dans un navigateur web)
- Le résultat est sous forme de JSON, donc de listes et de de dictionnaires imbriqués.

Prenons l'exemple d'une API qui répertorie des blagues sur Chuck Norris. Tu peux accéder à [ce premier lien](https://api.chucknorris.io/jokes/random) depuis ton navigateur. C'est ce qu'on appelle le **endpoint**, le point d'accès de l'API. Tu devrais obtenir quelque chose comme ceci :

```shell
{"categories":[],"created_at":"2020-01-05 13:42:29.855523","icon_url":"https://api.chucknorris.io/img/avatar/chuck-norris.png","id":"h8Ug7JjlTYSAoLAwngZWrA","updated_at":"2020-01-05 13:42:29.855523","url":"https://api.chucknorris.io/jokes/h8Ug7JjlTYSAoLAwngZWrA","value":"When Chuck Norris claps his hands, the sun turns off."}
...
```

C'est un format JSON, c'est-à-dire des listes et des dictionnaires imbriqués.

### Et en Python ?

Avec Python, pour effectuer des requêtes depuis le protocole HTTP, le plus simple est de passer par la bibliothèque `requests`.

```python
import requests
link = 'https://api.chucknorris.io/jokes/random'
r = requests.get(link)
print(r)

```

Le résultat est un code `200`, cela signifie que le protocole HTTP a bien transféré des données

```shell-black
>>> <Response [200]>
```

Pour accéder au contenu, tu peux utiliser la méthode `.json()` :

```python
import requests
link = 'https://api.chucknorris.io/jokes/random'
r = requests.get(link)
print(r.json())
```

```shell-black
>>> {'categories': [], 'created_at': '2020-01-05 13:42:30.730109', 'icon_url': 'https://api.chucknorris.io/img/avatar/chuck-norris.png', 'id': '_eBhT44NSq28ErFpLyB1vQ', 'updated_at': '2020-01-05 13:42:30.730109', 'url': 'https://api.chucknorris.io/jokes/_eBhT44NSq28ErFpLyB1vQ', 'value': 'Chuck Norris went to Ihop for pizza'}
```

1. Lire la documentation de l'API pour affiner la requête

### Lire la documentation

La "requête" commence donc :
- par un **endpoint** (une URL à requêter en HTTP)
- suivi d'un point d'interrogation `?`
- suivi de filtres sur des variables

Les filtres sur les variables dépendent de l'implémentation de celui qui développe l'API. Il n'y a pas de règle unique.


Pour cette partie, nous allons essayer une **autre API** qui permet de récupérer des informations sur les **chiens**. Nous allons utiliser **The Dog API**. La documentation de cette API est disponible [ici](https://dog.ceo/dog-api/).

Dans cette API, nous pouvons utiliser la variable `breed`. Le détail de la documentation indique : *Returns a random dog image from the breed specified in the query parameter*.

### Exemple d'utilisation de l'API

Essayons de l'utiliser. Chaque variable doit :
- être indiquée sans caractère espace
- être suivie d'un caractère égal `=`
- suivi de la valeur du filtre



````python
link = 'https://dog.ceo/api/breeds/image/random?breed=bulldog'
r = requests.get(link)
print(r.json())
````

Si tu souhaites utiliser plusieurs filtres, tu sépares les filtres par des esperluettes `&`.


````python
link = 'https://dog.ceo/api/breeds/image/random/3?breed=bulldog&breed=lama'
r = requests.get(link)
print(r.json())
````
